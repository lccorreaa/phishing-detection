#### Variables
-- --
- sender - This is the email of the sender of the email, in the format *"Name <<email@email.com>>"* or just "email@email.com"  

- receiver - This is the email of the person receiving the email, the recipient, in the format *"Name <<email@email.com>>"* or just "email@email.com"   
- date - Date the email was sent from sender to recipient, in the format *"Tue, 05 Aug 2008 19:31:21 -0400"*  
- subject - Subject line of the email-- Title of email   
- body - Main content of the email  
- label - Spam vs legitimate label, spam emails have a label of 1, normal emails have a label of 0  
- url -  1 if the body has any links/urls and 0 otherwise
